```python
# Netflix Data Analysis Project
# Author: DILKHUS KUMAR DAS

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("netflix_data.csv")

# Display first few rows
print("===== Netflix Data Head =====")
print(df.head())

# Check for missing values
print("\n===== Missing Values =====")
print(df.isnull().sum())

# Drop duplicates
df.drop_duplicates(inplace=True)

# Data Overview
print("\n===== Data Info =====")
print(df.info())

# Count of Movies vs TV Shows
type_count = df['type'].value_counts()
print("\n===== Count of Movies vs TV Shows =====")
print(type_count)

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='type', palette='coolwarm')
plt.title("Movies vs TV Shows on Netflix")
plt.show()

# Top 5 Countries with Most Titles
plt.figure(figsize=(8, 5))
df['country'].value_counts().head(5).plot(kind='bar', color='skyblue')
plt.title("Top 5 Countries with Most Netflix Titles")
plt.xlabel("Country")
plt.ylabel("Count")
plt.show()

# Most common genres
plt.figure(figsize=(10, 5))
genres = df['listed_in'].value_counts().head(5)
sns.barplot(x=genres.values, y=genres.index, palette='mako')
plt.title("Top 5 Netflix Genres")
plt.xlabel("Count")
plt.ylabel("Genre")
plt.show()

# Release Year Trend
plt.figure(figsize=(10, 5))
sns.histplot(df['release_year'], bins=10, kde=True)
plt.title("Number of Releases per Year")
plt.xlabel("Release Year")
plt.ylabel("Count")
plt.show()

# Rating Distribution
plt.figure(figsize=(8, 5))
sns.countplot(data=df, y='rating', order=df['rating'].value_counts().index, palette='Set2')
plt.title("Distribution of Content Ratings")
plt.xlabel("Count")
plt.ylabel("Rating")
plt.show()

print("\n===== Insights =====")
print("1️⃣ Movies dominate over TV Shows.")
print("2️⃣ Most Netflix titles come from the United States and India.")
print("3️⃣ Popular genres include Drama, Action, and Sci-Fi.")
print("4️⃣ Netflix content production has increased rapidly after 2015.")
print("5️⃣ The majority of content is rated for mature audiences (TV-MA or R).")
```
